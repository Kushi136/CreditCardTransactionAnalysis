# Credit Card Transaction Analysis & Fraud Detection
## Stage 2: Exploratory Data Analysis (EDA) & Visualizations

This notebook explores patterns, distributions, class imbalance, temporal spikes, merchant categories, customer demographics, and geospatial behaviors associated with fraudulent transactions.

### Step 1: Load Cleaned Dataset

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
clean_train_path = os.path.join("..", "data", "processed", "fraudTrain_clean.csv")
train = pd.read_csv(clean_train_path)
train.head()
```

### EDA 1 & 2: Class Imbalance (Fraud vs Genuine)

```python
print("Counts:")
print(train["is_fraud"].value_counts())
print("\nPercentages:")
print(train["is_fraud"].value_counts(normalize=True) * 100)

plt.figure(figsize=(7, 5))
sns.countplot(data=train, x="is_fraud", palette="coolwarm")
plt.title("Fraud vs Genuine Transactions")
plt.xlabel("Transaction Type (0 = Genuine, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.show()
```

### EDA 3: Transaction Amount Distribution & Boxplot

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=train[train["amt"] < 1000], x="amt", bins=50, kde=True)
plt.title("Transaction Amount Distribution (< $1,000)")
plt.xlabel("Transaction Amount ($)")
plt.ylabel("Number of Transactions")
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=train, x="is_fraud", y="amt", palette="Set2", showfliers=False)
plt.title("Transaction Amount: Fraud vs Genuine")
plt.show()
```

### EDA 4: Transactions by Merchant Category

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=train, x="category", hue="is_fraud", palette="Set1")
plt.xticks(rotation=45, ha="right")
plt.title("Transactions by Category")
plt.show()
```

### EDA 5: Fraud by Transaction Hour

```python
plt.figure(figsize=(12, 5))
sns.countplot(data=train[train["is_fraud"] == 1], x="transaction_hour", palette="viridis")
plt.title("Fraud Transactions by Hour")
plt.xlabel("Hour of Day (0-23)")
plt.ylabel("Count")
plt.show()
```

### EDA 6: Fraud by Day of Week

```python
day_names = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
plt.figure(figsize=(10, 5))
sns.countplot(data=train[train["is_fraud"] == 1], x="day_of_week", palette="magma")
plt.title("Fraud by Day of Week")
plt.xticks(ticks=range(7), labels=day_names)
plt.xlabel("Day of Week")
plt.show()
```

### EDA 7: Fraud by Gender

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=train, x="gender", hue="is_fraud", palette="pastel")
plt.title("Fraud by Gender")
plt.show()
```

### EDA 8: Top States with Highest Fraud

```python
fraud_states = train[train["is_fraud"] == 1]["state"].value_counts().head(10)
print("Top 10 States by Fraud:")
print(fraud_states)

fraud_states.plot(kind="bar", figsize=(10, 5), color="#d9534f")
plt.title("Top States by Fraud Transactions")
plt.xlabel("State")
plt.ylabel("Fraud Transactions")
plt.show()
```

### EDA 9: Customer Age Distribution

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=train, x="customer_age", hue="is_fraud", bins=30, kde=True)
plt.title("Customer Age Distribution by Fraud Status")
plt.show()
```
